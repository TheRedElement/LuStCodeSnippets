## `add_hovergrid()`

```python
importlib.reload(plyc)

x = np.linspace(-2, 2, 100)
y = x**2
fig = make_subplots(1,1, subplot_titles=["hover to see additional info"])
fig.add_trace(go.Scatter(x=x, y=y, mode="markers"), row=1, col=1)
fig.add_trace(go.Scatter(x=x, y=np.sqrt(y), mode="markers"))

xx, yy = np.meshgrid(x, y)
xx, yy = xx.flatten(), yy.flatten()

fig = plyc.add_hovergrid(fig, row=1, col=1,
    x=xx, y=yy,
    debug=False,
    custom1=np.round(np.abs(xx),3), custom2=np.random.choice(["Text1", "Text2", "Text3"], replace=True, size=len(xx))
)
fig.show()
```
