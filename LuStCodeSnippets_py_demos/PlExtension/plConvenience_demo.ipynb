```python
#example for a histogram (binning of column_2 w.r.t. column_0)
df_hist = (plc.cut(df, "column_0", [2,3])
    .group_by("column_0_cut").agg(pl.col("column_2").mean())
    .sort(pl.col("column_0_cut"))
)
display(df_hist)

fig, axs = plt.subplots(1,1, subplot_kw=dict(xlabel="cut", ylabel="mean(column_2)"))
axs.bar(df_hist["column_0_cut"], df_hist["column_2"])
plt.show()
```
