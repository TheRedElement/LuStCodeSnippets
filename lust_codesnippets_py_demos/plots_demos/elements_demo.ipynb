## `pcolormesh_text()`

```python
X = np.random.randn(5,7)
X[1,3] = np.nan

fig = plt.figure()
ax = fig.add_subplot(111)
mesh = ax.pcolormesh(X, vmin=-0.5, vmax=0.5)
pelements.pcolormesh_text(ax,
    X,
    cmap=None, nancolor="r",
    norm=mcolors.Normalize(vmin=-0.5, vmax=0.5),
    numformat="%.2f",
    xoffset=0.5, yoffset=0.5,
    text_kwargs=None,
)
cbar = fig.colorbar(mesh, ax=ax)
plt.show()
```

## `legend_line_map()`

```python

nlines = 5
ls = ["-", "--", "-.", ":", "-"]
lw = np.linspace(0.1,2,nlines)
fig, axs = plt.subplots(2,1)
x = np.linspace(0,1,100)
y = np.array([x**i for i in range(nlines)])

for idx, yi in enumerate(y):
    axs[0].plot(x, yi, ls=ls[idx], label=f"x**{idx}->{idx}")
    axs[1].plot(x, np.sqrt(yi), lw=lw[idx], label=f"sqrt(x**{idx})->{idx}_2")

handles, labels = pelements.get_handles_labels_all(fig)
pelements.legend_line_map(
    fig,
    handles=handles, labels=labels,
    header="Function->Exponent",
    x0=0.02, x1=0.4, y0=0.9,
    vsep=1.0, pad=0.01,    
    topdown=True,
    text_kwargs=dict(color="red"),
    header_kwargs=dict(fontsize=13)
)
plt.show()
```
