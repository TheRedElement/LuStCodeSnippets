```python
#%%imports
import pytest
from lust_codesnippets_py.sklearn_extension import clfmetric_error as clfe

import numpy as np
```

## Observed quantities
* TP ... true positives
* TN ... true negatives
* FP ... false positives
* FN ... false negatives

```python
obs = np.array([5,5,5,5])   #(TP,TN,FP,FN)
obs_e = 1/np.sqrt(obs)
```

## Accuracy

$$
\begin{align}
	Acc
		&= \frac{TP + TN}{TP+TN+FP+FN}
\end{align}
$$

```python
acc = np.sum(obs[[0,1]])/np.sum(obs)
acc_e = clfe.clfmetric_error(obs, obs_e, [0,1])
print(f"Acc = {acc} +/- {acc_e}")
```

## Contamination

$$
\begin{align}
	Cont
		&= \frac{FP}{TP+FP}
\end{align}
$$

```python
cont = np.sum(obs[2])/np.sum(obs[[0,2]])
cont_e = clfe.clfmetric_error(obs, obs_e, [1])
print(f"Cont = {cont} +/- {cont_e}")
```

## Recall

$$
\begin{align}
	Rec
		&= \frac{TP}{TP+FN}
\end{align}
$$

```python
rec = np.sum(obs[0])/np.sum(obs[[0,3]])
rec_e = clfe.clfmetric_error(obs, obs_e, [1])
print(f"Rec = {rec} +/- {rec_e}")
```
